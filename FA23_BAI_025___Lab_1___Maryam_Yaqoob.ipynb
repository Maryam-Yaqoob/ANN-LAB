## Task 6: Conditional Statements
*Topics: Flow Control (if / elif / else)*

**a)** Grade from marks

```python
marks = 82

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Marks: {marks} -> Grade: {grade}")
```

**b)** Positive, negative or zero

```python
num = -7

if num > 0:
    print(num, "is positive")
elif num < 0:
    print(num, "is negative")
else:
    print("The number is zero")
```

**c)** Leap year using nested if

```python
year = 2024

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
        print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

**d)** Scholarship eligibility with `and` / `or`

```python
cgpa = 3.7
attendance = 85
has_financial_need = False

if cgpa > 3.5 and attendance > 80:
    print("Eligible for merit scholarship")
elif cgpa > 3.0 or has_financial_need:
    print("Eligible for partial/need-based scholarship")
else:
    print("Not eligible for a scholarship")
```

## Task 7: Loops with Break and Continue
*Topics: Flow Control (for / while / break / continue)*

**a)** Print 1 to 30, skipping multiples of 4

```python
for i in range(1, 31):
    if i % 4 == 0:
        continue
    print(i, end=" ")
```

**b)** Print 1 to 50, stop when a number greater than 20 is found

```python
for i in range(1, 51):
    if i > 20:
        print("\nStopped at", i)
        break
    print(i, end=" ")
```

**c)** Multiplication table using a while loop

```python
n = 7
i = 1
while i <= 10:
    print(f"{n} x {i} = {n * i}")
    i += 1
```

**d)** Sum of even numbers from 1 to 100

```python
total = 0
for i in range(1, 101):
    if i % 2 == 0:
        total += i
print("Sum of even numbers (1-100):", total)
```

## Task 9: Classes and Inheritance
*Topics: Classes (OOP Basics)*

**a)** `Course` class

```python
class Course:
    def __init__(self, course_name, credit_hours):
        self.course_name = course_name
        self.credit_hours = credit_hours

    def display_info(self):
        print(f"Course: {self.course_name}, Credit Hours: {self.credit_hours}")
```

**b)** Two `Course` objects

```python
c1 = Course("Programming Fundamentals", 3)
c2 = Course("Calculus", 4)

c1.display_info()
c2.display_info()
```

**c)** `LabCourse` inheriting from `Course`

```python
class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours
```

**d)** Override `display_info()`

```python
class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours

    def display_info(self):
        super().display_info()
        print(f"Lab Hours: {self.lab_hours}")

lab = LabCourse("Physics Lab", 1, 3)
lab.display_info()
```
