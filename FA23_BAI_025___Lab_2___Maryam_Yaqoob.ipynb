```python
import numpy as np
```

## Task 1: Vectors

```python
ages = np.array([22, 35, 41, 28, 53, 60])
print("Ages:", ages)
print("Shape:", ages.shape)
print("Dimensions:", ages.ndim)
print("Data type:", ages.dtype)

# Years until retirement (vectorised, no loop)
years_to_retirement = 65 - ages
print("Years until retirement:", years_to_retirement)

# Average age using a NumPy function
print("Average age:", np.mean(ages))
```

## Task 2: Matrices

```python
sales = np.array([[120, 135, 150],
                  [ 80,  95, 110],
                  [200, 180, 210],
                  [ 60,  75,  90]])
print(sales)
print("Shape:", sales.shape)
print("Total elements:", sales.size)

zero_matrix = np.zeros(sales.shape)
print("\nZero matrix:\n", zero_matrix)

# Replace second row using indexing
zero_matrix[1] = [10, 20, 30]
print("\nAfter replacing second row:\n", zero_matrix)
```

## Task 3: Addition & Subtraction

```python
term1 = np.array([[70, 65, 80],
                  [55, 60, 70],
                  [90, 85, 75]])
term2 = np.array([[75, 70, 85],
                  [65, 72, 78],
                  [88, 90, 80]])

total_marks = term1 + term2
print("Total marks (Term1 + Term2):\n", total_marks)

improvement = term2 - term1
print("\nImprovement (Term2 - Term1):\n", improvement)

# Total improvement per student (sum across subjects, axis=1)
total_improvement = improvement.sum(axis=1)
print("\nTotal improvement per student:", total_improvement)

best_student = np.argmax(total_improvement)
print(f"Student {best_student + 1} had the highest total improvement: {total_improvement[best_student]}")
```

## Task 4: Dot Product

```python
prices = np.array([50, 120, 30, 200])
quantities = np.array([2, 1, 5, 3])

total_bill = np.dot(prices, quantities)
print("Total bill:", total_bill)

# 2 customers x 4 items
customers_qty = np.array([[2, 1, 5, 3],
                          [1, 0, 2, 4]])

# (2x4) . (4,) -> (2,) : one bill per customer
bills = np.dot(customers_qty, prices)
print("Bills for both customers:", bills)

# Why shapes must match:
# The dot product multiplies matching elements and sums them, so the inner dimensions
# must be equal. For (2x4) . (4,), the 4 columns (items) of the quantity matrix line up
# with the 4 prices. Each row (customer) is paired item-by-item with the price vector.
# If the sizes differed (e.g. 3 prices vs 4 quantities), some elements would have no
# partner, and NumPy raises a ValueError.
```

## Task 5: Reshaping

```python
hours = np.array([2, 3, 1, 4, 5, 2, 3, 4, 6, 1, 2, 3])
print("Original:", hours, hours.shape)

# 4 rows x 3 columns (4 weeks of a 3-day study plan)
reshaped_4x3 = hours.reshape(4, 3)
print("\n4x3:\n", reshaped_4x3)

# A different valid shape
reshaped_2x6 = hours.reshape(2, 6)
print("\n2x6:\n", reshaped_2x6)

# Invalid reshape
try:
    hours.reshape(5, 3)
except ValueError as e:
    print("\nError:", e)

# Error message:
# ValueError: cannot reshape array of size 12 into shape (5,3)
```

## Task 8: Random Matrices

```python
np.random.seed(42)

# 3x3 random decimals between 0 and 1
rand_decimals = np.random.rand(3, 3)
print("Random decimals:\n", rand_decimals)

# 3x3 random integers between 10 and 100 (upper bound is exclusive)
rand_ints = np.random.randint(10, 101, size=(3, 3))
print("\nRandom integers:\n", rand_ints)

# 4x4 standard normal
rand_normal = np.random.randn(4, 4)
print("\nStandard normal:\n", rand_normal)
```

```python
# Run part 1 twice using the same seed
np.random.seed(42)
run1 = np.random.rand(3, 3)

np.random.seed(42)
run2 = np.random.rand(3, 3)

print("Run 1:\n", run1)
print("Run 2:\n", run2)
print("Identical:", np.array_equal(run1, run2))

# Why identical: random numbers from NumPy are pseudo-random. They come from a
# deterministic algorithm whose starting state is set by the seed. The same seed gives
# the same starting state, hence the same sequence of numbers every time.
# (If the seed were NOT reset between the two calls, the generator would continue
# from where it left off and the second output would be different.)
```

## Task 9: Transpose

```python
scores = np.array([[80, 85, 90, 78, 88],
                   [70, 75, 72, 80, 77],
                   [95, 92, 89, 94, 96]])
print("Original shape:", scores.shape)

scores_T = scores.T
print("Transposed shape:", scores_T.shape)
print(scores_T)

# Transposing twice returns the original
print("\nTranspose twice equals original:", np.array_equal(scores_T.T, scores))
```

```python
A = np.array([[1, 2],
              [3, 4],
              [5, 6],
              [7, 8]])        # 4x2
B = np.array([[1, 0],
              [2, 1],
              [0, 3]])        # 3x2

# A (4x2) . B (3x2) is not possible: inner dimensions 2 and 3 do not match.
# Transposing B gives 2x3, so (4x2) . (2x3) works and gives 4x3.
result = np.dot(A, B.T)
print("Result shape:", result.shape)
print(result)
```
